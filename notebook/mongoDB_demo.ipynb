```python

import pandas as pd
import pymongo

df = pd.read_csv('data.csv')
df.head()
df.shape
```

```python

# df should be converted into dict before we push it to mongodb

data = df.to_dict(orient='records')
# data
```

```python
DB_NAME = "proj1"
COLLECTION_NAME = "Proj1-Data"
CONNECTION_URL = "mongodb+srv://suyashv560_db_user:lnwITDU9ag47dmPR@cluster0.oaswink.mongodb.net/"
```

```python
client = pymongo.MongoClient(CONNECTION_URL)
data_base = client[DB_NAME]
collection = data_base[COLLECTION_NAME]
```

```python
client.admin.command("ping")
print("MongoDB connected successfully!")
```

```python
# Upload data to MongoDB in batches
batch_size = 5000

for i in range(0, len(data), batch_size):
    batch = data[i:i + batch_size]

    collection.insert_many(batch, ordered=False)

    print(f"Uploaded {min(i + batch_size, len(data))}/{len(data)}")

print("Upload completed!")
```
